```python
import numpy as np
import matplotlib.pyplot as plt
font = {"size": 26}

plt.rc("font", **font)

plt.rcParams["figure.figsize"] = (16, 9)
plt.rcParams["lines.linewidth"] = 4
```

```python
I_oa=np.genfromtxt("phi_r_OA.csv", delimiter=",")
```

```python
steps = [0,50,100,250,500,1000]
```

```python
f_targ = 0.05
dz = 0.02
nzf = 501

zi = f_targ - dz
zf = f_targ + dz
zF = np.linspace(zi, zf, nzf)
```

```python
np.shape(I_oa)
```

```python
len(I_oa[:,0])
```

```python
target = np.genfromtxt('../../../I_oa_para.csv')
```

```python
plt.plot((zF-f_targ) * 100, target, c='k', label = r'$I_T$')

line_colors = plt.cm.viridis_r(np.linspace(0, 1, len(steps)))

for i,num in enumerate(steps):
    plt.plot((zF-f_targ) * 100, I_oa[num], alpha = 0.7, c=line_colors[i], label = r"$I_R$"' step %d' % num)

    
plt.ylabel("On-axis Intensity ")
plt.xlabel(r"$z-f$ (cm)")
plt.ticklabel_format(axis="y", style="sci", scilimits=(0, 0))
# plt.xlim(-3,7)
plt.legend()
plt.title('Reflected Intensity over Iterations')
#     plt.show()
plt.tight_layout()
plt.savefig('I_r_oa_over_iterations')
```
