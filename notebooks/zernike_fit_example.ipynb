Fits Zernike coefficients to a sampled wavefront `W(rho, theta)` via least squares (`(A^T A)^-1 A^T W`), following the worked example from:

[Zernike-Fit.pdf](https://wp.optics.arizona.edu/visualopticslab/wp-content/uploads/sites/52/2021/10/Zernike-Fit.pdf) (University of Arizona Visual Optics Lab)

Citation (in case the link above goes dead — the PDF's embedded metadata identifies it as an excerpt of this book):

J. Schwiegerling, *Optical Specification, Fabrication, and Testing*, SPIE Press, Bellingham, WA (2014).

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
rho=np.array([1.  , 0.5 , 0.25, 0.5 , 1.  , 0.25, 0.  , 0.25, 1.  , 0.5 , 0.25, 0.5 , 1.  ])
theta=np.array([ np.pi/2,  3*np.pi/4,  np.pi/2,  np.pi/4,  np.pi,  np.pi, 0,  0.,  0.,  -3*np.pi/4, -np.pi/2, -np.pi/4,-np.pi/2])
W=0.25+2*np.sqrt(3)*(2*rho**2-1)+0.5*np.sqrt(6)*rho**2*np.cos(2*theta)
```

```python
def zern_pols(rho,theta):
    # first 6 Zernike polynomials
    pols=[np.ones_like(rho), 2*rho*np.sin(theta), 2*rho*np.cos(theta), np.sqrt(6)*rho**2*np.sin(2*theta), np.sqrt(3)*(2*rho**2-1), np.sqrt(6)*rho**2*np.cos(2*theta)]
    # pols=[np.ones_like(rho), np.sqrt(3)*(2*rho**2-1), np.sqrt(5) * (6 * rho**4 - 6 * rho**2 + 1)]

    return np.column_stack(pols)  
```

```python
A=zern_pols(rho,theta)
#print each column of A
for i in range(3):
    print(A[:,i])
```

```python
ATA_inv=np.linalg.inv(A.T @ A)
print(ATA_inv)
```

```python
x=ATA_inv @ A.T @ W
print(x)
```
