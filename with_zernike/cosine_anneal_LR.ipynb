```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt

# --- Formatting (matches plot.py style) ---
font = {"size": 30}
plt.rc("font", **font)
plt.rcParams["figure.figsize"] = (16, 9)
plt.rcParams["lines.linewidth"] = 4
mpl.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": ["Computer Modern Roman"],
})

# Parameters
eta_max_init = 0.1    # initial max LR
eta_min = 0.001       # min LR
T_0 = 200             # initial cycle length
num_steps = 200*(2**6-1)    # total steps
cycle_mult = 2        # cycle length multiplier
decay_factor = 1      # decay eta_max after each restart

# Generate learning rate schedule
eta = []
t_cur = 0
T_cur = T_0
eta_max = eta_max_init

for t in range(num_steps):
    eta_t = eta_min + 0.5 * (eta_max - eta_min) * (1 + np.cos(np.pi * t_cur / T_cur))
    eta.append(eta_t)

    t_cur += 1
    if t_cur >= T_cur:
        t_cur = 0
        T_cur *= cycle_mult
        eta_max *= decay_factor

# Plot
plt.figure()
plt.plot(range(num_steps), eta)
plt.xlabel(r"Step")
plt.ylabel(r"Learning Rate")
plt.title(r"Cosine Annealing with Warm Restarts")
plt.tight_layout()
plt.show()
```
